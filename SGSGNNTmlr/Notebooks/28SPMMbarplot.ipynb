```python
import matplotlib.pyplot as plt
import numpy as np

# Data structured as: {Dataset: {'edge': (SPMM, DDMM), 'full': (...), 'learned': (...)}}
data = {
    'amherest': {
        'edge': (67.02248238, 32.97751762),
        'full': (73.36780733, 26.63219267),
        'learned': (70.35565804, 29.64434196),
    },
    'SmallCora': {
        'edge': (74.01375149, 25.98624851),
        'full': (74.69616838, 25.30383162),
        'learned': (67.74789087, 32.25210913),
    },
    'Arxiv-Year': {
        'edge': (71.88895926, 28.11104074),
        'full': (72.43589634, 27.56410366),
        'learned': (76.09143555, 23.90856445),
    },
    'Reddit': {
        'edge': (61.62018526, 38.37981474),
        'full': (71.70561812, 28.29438188),
        'learned': (72.13139792, 27.86860208),
    }
}

methods = ['edge', 'full', 'learned']
colors = {'SPMM': 'red', 'DDMM': 'blue'}

fig, ax = plt.subplots(figsize=(12, 6))
bar_width = 0.2
group_spacing = 0.6
method_spacing = 0.05

x_labels = []
x_ticks = []
x_pos = 0

for dataset, values in data.items():
    for i, method in enumerate(methods):
        spmm, ddmm = values[method]
        ax.bar(x_pos, spmm, color=colors['SPMM'], width=bar_width, label='SPMM' if x_pos == 0 else "")
        ax.bar(x_pos, ddmm, bottom=spmm, color=colors['DDMM'], width=bar_width, label='DDMM' if x_pos == 0 else "")

        x_labels.append(f"{dataset}\n{method}")
        x_ticks.append(x_pos)
        x_pos += bar_width + method_spacing  # space between methods
    x_pos += group_spacing  # extra space between dataset groups

ax.set_xticks(x_ticks)
ax.set_xticklabels(x_labels, rotation=45, ha='right')
ax.set_ylabel("Percentage (Total = 100)")
ax.set_title("SPMM vs DDMM for each method across datasets")
ax.legend()
plt.tight_layout()
plt.show()
```
